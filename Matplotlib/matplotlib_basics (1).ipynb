```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
students = ["A", "B", "C", "D"]
marks = [85, 90, 78, 92]

plt.bar(students, marks)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
data = [10, 20, 20, 30, 30, 30, 40, 40, 50]

plt.hist(data)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 6, 8, 9]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
subjects = ["Python", "ML", "DL", "SQL"]
hours = [30, 25, 20, 15]

plt.pie(hours, labels=subjects, autopct="%1.1f%%")
plt.title("Study Time Distribution")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.show()
```
