```python
import numpy as np

arr1 = np.array ([10,20,30,40,50])
```

```python
print ("Array: ")
print (arr1)

print ("\nShape : ",arr1.shape)
print("Dimensions: ", arr1.ndim )
print("Data Type :", arr1.dtype)
print ("Size" , arr1.size)
print ("Item Size: ", arr1.itemsize)
```

```python
#indexing and slicing
print( "first element :" ,arr1[0] )
print( "last element :", arr1[-1]  )
print(   "slice :" , arr1[1:4])

#reshaping
matrix =np.arrange(1,13).reshape(3,4)
print(matrix)

print("Rows:", matrix.shape[0])
print("Columns:", matrix.shape[1])
```

```python
#statistical operations

data = np.array([25, 30, 35, 40, 45])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Maximum:", np.max(data))
print("Minimum:", np.min(data))
```

```python
#vectorized operations
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Square:", a ** 2)
```

```python
#random data
np.random.seed(42)

random_data = np.random.randint(1, 100, size=10)

print(random_data)
```

```python
#filtering data
scores = np.array([45, 78, 88, 92, 56, 39, 95])

passed = scores[scores >= 60]

print("Passed Students:")
print(passed)
```

```python
study_hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([35, 45, 50, 60, 70, 85])

correlation = np.corrcoef(study_hours, marks)

print(correlation)
```

```python
#mini data analysis
import numpy as np

np.random.seed(42)

study_hours = np.random.randint(1, 11, 50)

marks = 30 + study_hours * 6 + np.random.randint(-10, 10, 50)

print("Study Hours:", study_hours)
print("Marks:", marks)

print("Average Study Hours:", np.mean(study_hours))
print("Average Marks:", np.mean(marks))

correlation = np.corrcoef(study_hours, marks)
print("Correlation:")
print(correlation)
```
