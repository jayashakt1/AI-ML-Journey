```python
import pandas as pd
```

```python
#creating dataframe
data = {
    "Name": ["Jaya", "Rahul", "Priya", "Amit"],
    "Marks": [85, 92, 78, 88],
    "Attendance": [90, 95, 88, 91]
}

df = pd.DataFrame(data)

df
```

```python
#basic information
print (df.shape)
print (df.columns)
print (df.dtypes)

#statistics
df.describe()
```

```python
#sorting
df.sort_values("Marks", ascending=False)
```

```python
#filtering
high_scorers = df[df["Marks"] > 80]

high_scorers
```

```python
#create new column
df["Result"] = df["Marks"] >= 80

df
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

```python
#lambda function
df["Performance"] = df["Marks"].apply(
    lambda x: "Excellent" if x >= 90
    else "Good" if x >= 80
    else "Average"
)

df
```
